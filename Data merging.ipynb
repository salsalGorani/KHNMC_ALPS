```python
# Defines
def analyse_df(df):
    nan_cnt = df.isnull().sum()
    nan_rto = df.isnull().mean()*100
    unique_vals_cnt = df.nunique()

    ana_result = pd.DataFrame({
        'NaN Counts' : nan_cnt
        , 'NaN Ratio (%)' : nan_rto
        , 'Unique Values' : unique_vals_cnt
    })
    return ana_result
```

```python
def merge_pipeline(b1df, b2df):
    # key column to --> idx
    b1df = b1df.set_index('PID')
    b2df = b2df.set_index('PID')

    # Col mapping
    mapping = {
        'Age': 'Age',
        'Gender': 'Sex',
        'MMSE': 'MMSE',
        'TIV': 'TIV',
        'global GMV': 'global GMV',
        'global WMV': 'global WMV',
        'global CSFV': 'global CSFV',
        'DDX': '최종진단 DDX'
    }
    for tgt, src in mapping.items():
        b1df[tgt] = b2df[src]

    # Column init
    b1df['Group'] = ''

    # col 'Sex' (col 'Gender' → F:0 / M:1)
    b1df['sex'] = b1df['Gender'].map({'F': 0, 'M': 1})

    # Add cols (only exists in DF 'b2*')
    extras = ['CDR', 'CDRSB', 'GDS', 'Barthel A DL', 'Instrumental A DL', 'SGDS', 'NPI']
    for col in extras:
        b1df[col] = b2df.get(col)

    # Cols sorting: insert col 'extras' behind col 'MMSE'
    cols = list(b1df.columns)
    for col in extras:
        cols.remove(col)
    insert_at = cols.index('MMSE') + 1
    for col in reversed(extras):
        cols.insert(insert_at, col)
    b1df = b1df[cols]

    # Reset idx
    return b1df.reset_index()
```

```python
PATTERNS = {
    'CN' : re.compile(r'\b(NL|normal\s*|cognitive\s*normal)\b', flags=re.I),

    'MCI': re.compile(
        r'\b('
        r'MCI|'                  # 단독 MCI
        r'aMCI|'                 # amnestic MCI
        r'non[-\s]?amnestic\s*MCI|' 
        r'multiple\s*domain\s*a?MCI|' 
        r'single\s*domain\s*a?MCI|' 
        r'mutiple\s*domain.*MCI'  # 오타 포함
        r')\b', flags=re.I),

    'AD' : re.compile(
        r'\b('
        r'AD\b|'                                     # AD
        r'Alzheimer|\bADD\b|'                        # Alzheimer variants
        r'알츠하이머형\s*치매|'
        r'dementia'                                  # any “dementia … AD”류
        r')', flags=re.I)
}

def classify_ddx(ddx):
    """Col 'DDX' values --> Col 'Group' CN / MCI / AD, Matching X --> None"""
    if not isinstance(ddx, str):
        return None
    for label, pat in PATTERNS.items():
        if pat.search(ddx):
            return label
    return None

for df in (b1AD, b1CN):
    df['Group'] = df['DDX'].apply(classify_ddx)
print(b1AD['Group'].value_counts(dropna=False))
print(b1CN['Group'].value_counts(dropna=False))
```

```python
hfc = hfc.rename(columns={'ID': 'PID'})
hfc['PID'] = (hfc['PID']
                .astype(str)
                .str.replace(r'\.0$', '', regex=True)
                .str.zfill(8))
print(hfc[['PID']])

roi_prefix_map = {
    'ROI1':  'Conduct1(age+)',
    'ROI2':  'Conduct2(age+)',
    'ROI3':  'Conduct1(age-)',
    'ROI4':  'Conduct1(MMSE+)',
    'ROI5':  'Conduct1(MMSE-)',
    'ROI6':  'Conduct2(MMSE-)',
    'ROI7':  'Conduct1(CN<MCI)',
    'ROI8':  'Conduct2(CN<MCI)',
    'ROI9':  'Conduct1(CN>AD)',
    'ROI10': 'Conduct1(CN<AD)',
    'ROI11': 'Conduct2(CN<AD)',
}

def rename_roi(col: str) -> str:
    """
    ROI9_WMV  →  Conduct1(CN>AD)_WMV
    접미사(_WMV 등)는 그대로 두고 접두사만 바꿔 준다
    """
    m = re.match(r'^(ROI\d{1,2})(_.+)$', col)
    if m and m.group(1) in roi_prefix_map:
        return f"{roi_prefix_map[m.group(1)]}{m.group(2)}"
    return col

hfc.columns = [rename_roi(c) for c in hfc.columns]

# ROI 열만 따로 추출 (+ PID)
roi_cols = [c for c in hfc.columns if re.match(r'^Conduct', c)]
hfc_roi  = hfc[['PID', *roi_cols]].copy()

hfc_roi
```

```python
def attach_roi(target: pd.DataFrame, roi_source: pd.DataFrame) -> pd.DataFrame:
    out = target.copy()
    out['PID'] = (
        out['PID'].astype(str)
                  .str.replace(r'\.0$', '', regex=True)
                  .str.zfill(8)
    )
    for c in roi_cols:
        if c not in out.columns:
            out[c] = np.nan
    merged = out.merge(roi_source, on='PID', how='left', suffixes=('', '_h'))

    for c in roi_cols:
        h_col = f'{c}_h'
        if h_col in merged.columns:
            out[c] = out[c].combine_first(merged[h_col])

    h_cols_to_drop = [col for col in out.columns if col.endswith('_h')]
    out.drop(columns=h_cols_to_drop, inplace=True)

    return out


# 적용
dfCN = attach_roi(dfCN, hfc_roi)
dfAD = attach_roi(dfAD, hfc_roi)
```
